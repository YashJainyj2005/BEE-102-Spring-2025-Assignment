### Importing Libraries

```python
import pandas as pd
import random
```

### Function to build transition matrix given an input sequence

```python
# Base list containing A,C,G,T for easier access ahead
bases = ['A', 'C', 'G', 'T']

def build_markov_transition_matrix(sequence):
  # Made a trans_frequency named dictionary which will store the frequency of
  # each Base X-> Base Y type of transitions for all the 16 combinations of A,C,G,T->A,C,G,T
    trans_frequency = {base: {b: 0 for b in bases} for base in bases}

    # Filling the dictionary with the frequencies by looping over the given sequence
    for i in range(len(sequence) - 1):
        current_base = sequence[i]
        next_base = sequence[i + 1]
        if current_base in bases and next_base in bases: # Checking if it is a valid base from A,C,G,T only or not
            trans_frequency[current_base][next_base] += 1

    # Now to get the transition matrix we need to normalize the frequncies to get the probabilities
    transition_matrix = {}
    for base in bases:
        total = sum(trans_frequency[base].values())
        if total > 0:
            transition_matrix[base] = {b: trans_frequency[base][b] / total for b in bases}
        else:
            transition_matrix[base] = {b: 0.0 for b in bases}

    # In order to print the transition matrix in a representable way, we can use pandas DataFrame
    df = pd.DataFrame(transition_matrix).T
    df.columns.name = 'To'
    df.index.name = 'From'
    return df
```

### Function to generate a random sequence to test the transition matrix function

```python
# Now in order to show this transition matrix, we need some sequence
# Since we were not given any input sequence, we can randomly generate sequences by using the following function
def generate_random_sequence(length):
    # In order to simulate realistic behaviour, we can introduce some bias in generation of the random sequence
    # Like here if first base is A then there is higher chance of getting C,T than A,G and so on as shown below
    transitions = {
        'A': ['A'] * 2 + ['C'] * 3 + ['G'] * 2 + ['T'] * 3, # This is the list ['A','A,','C','C','C','G','G','T','T','T']
        'C': ['A'] * 3 + ['C'] * 2 + ['G'] * 3 + ['T'] * 2,
        'G': ['A'] * 3 + ['C'] * 3 + ['G'] * 2 + ['T'] * 2,
        'T': ['A'] * 2 + ['C'] * 3 + ['G'] * 3 + ['T'] * 2
    }
    # Randomly generate the first base
    first_base=random.choice(bases)
    sequence = first_base
    for i in range(length - 1):
        last = sequence[-1]
        next_base = random.choice(transitions[last])
        sequence+=next_base
    # Return the sequence as a string
    return sequence
```
