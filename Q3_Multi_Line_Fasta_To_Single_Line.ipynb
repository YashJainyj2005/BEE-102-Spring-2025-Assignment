## Q3 Multi-line sequence fasta to single-line fasta file
Q- Fasta file format is shown below. Many times, it is annoying to find sequences of interest because of fix-width format, i.e., if 60 character fix-width file is there then only except the header (starting with >) of the sequence each line will have 60 characters. So, if a sequence is of width 600, then it would be written in 10 lines. Your goal is convert this format to a format with header in a single line followed by its sequence in a single line.

```python
# List-free solution idea:-

# To convert multi-line fasta to single-line without using list data structure,
#  we can just read the sequence line by line and keep on writing each line in the
#  output file, as soon as we will reach the end of the sequence which is confirmed
#  if the line starts with '>' then we will just go to the next line using '\n'
# and write that line (starting with >) and start writing the next sequence from next line.

# By this way we do not need to store the sequence in a list since if the sequence
# is too big like human chromosomes

def flatten_fasta_stream(input_file, output_file):
    written = False # Just a flag to know whether we have started writing or not
    with open(input_file, 'r') as infl, open(output_file, 'w') as otfl:
        for line in infl:
            line = line.strip() # Removing any extra white spaces at the end
            if line.startswith(">"): # If starting with '>' it means a new sequence is starting
                if written: # If we have already written something it means it is not the first line so write '\n' to write new sequence from the next line
                    otfl.write("\n")
                otfl.write(line + "\n")
                written = True
            else:
                otfl.write(line)
        otfl.write("\n")

input_file_path="multiline_input.fasta"
output_file_path="singleline_output.fasta"
flatten_fasta_stream(input_file_path, output_file_path)
print("The single-line fasta output has been stored in file named singleline_ouput.fasta in the current directory")
```
